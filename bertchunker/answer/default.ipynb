## Set up the frozen-encoder baseline

The paths below assume the notebook is running from `answer/`.

```python
chunker = FinetuneTagger(
    os.path.join('..', 'data', 'chunker'),
    modelsuffix='.pt',
    trainfile=os.path.join('..', 'data', 'train.txt.gz'),
    batchsize=16,  # Match the baseline command-line default.
)
```

## Train once before decoding

This cell trains a new frozen-encoder baseline and replaces `../data/chunker.pt`. Back up an existing checkpoint first. Skip this cell only when reusing a checkpoint produced by the corrected starter. Training time depends on your device.

The original pretrained encoder and tokenizer may be downloaded on first use. Review any library warning rather than assuming every warning is harmless.

```python
chunker.train()
```

## Run the frozen-encoder baseline on dev

```python
decoder_output = chunker.decode(os.path.join('..', 'data', 'input', 'dev.txt'))
```

## Evaluate the default output

```python
flat_output = [ output for sent in decoder_output for output in sent ]
sys.path.append('..')
import conlleval
true_seqs = []
with open(os.path.join('..', 'data', 'reference', 'dev.out')) as r:
    for sent in conlleval.read_file(r):
        true_seqs += sent.split()
conlleval.evaluate(true_seqs, flat_output)
```
